```python
#Write a Python script that asks a user for a temperature in Celsius and converts it to Fahrenheit.

tc = float(input("Enter temp in celsius:"))
tf = (tc*9/5)+32
print("Temp in fahrenheit",tf)
```

```python
#Write a program that looks at every number from 1 to 10 and prints whether each number is Even or Odd.

for i in range(1,11):
  if(i%2 == 0):
    print("Number is even", i)
  else:
    print("Number is odd", i)
```

```python
#Create a system that checks if a specific person is invited to a party.

guests = ["alab", "alya", "boi", "daina"]
x = str(input("enter a name:")).strip().title()
if x in guests:
  print("guests is invited")
else:
  print("guests is not invited")
```

```python
#Create a simple contact book where you can look up a friend's phone number.

contact = {
    "max" : "3892",
    "lily" : "8201",
    "omm" : "9371",
    "khushi" : "5291"
}
user_input = str(input("Enter the name:")).strip().lower()
if user_input in contact:
  print(contact[user_input])
else:
  print("Not found")

```

```python
#Write a function that calculates the area of a rectangle, and then use it to find the area of a room
def calculate_area(length, width):

    area = length * width
    return area


room_length = float(input("Enter room length: "))
room_width = float(input("Enter room width: "))


total_area = calculate_area(room_length, room_width)


print(f"The total area is: {total_area}")
```
